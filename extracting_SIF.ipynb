```python
import glob
import numpy as np
import rioxarray
import re
import pandas as pd

# Define coordinates
loc_coords = np.array([
    (41.1651, -96.4766),
    (41.1649, -96.4701),
    (41.1797, -96.4397),
    (38.4133, -120.9508)
])

# Step 1: Get files and sort by year, month
dir = '/Users/claireleeb/research/'
files = glob.glob(dir + '*.tif')
coords_used = []
def extract_date(f):
    match = re.search(r'(\d{4})\.M(\d{2})', f)
    if match:
        return int(match.group(1)), int(match.group(2))
    return (9999, 99)  # fallback

files_sorted = sorted(files, key=extract_date)

# Step 2: Extract SIF values
sif_values = np.zeros((len(files_sorted), len(loc_coords)))

for i, filename in enumerate(files_sorted):
    ds = rioxarray.open_rasterio(filename)
    for j, (lat, lon) in enumerate(loc_coords):
        dt = ds.sel(x=lon, y=lat, method="nearest").values[0]
        #sif_values[i,j] = dt
        #coords_used.append({
            #"loc": j,
            #"coordinates":
            
        #})
        sif_values[i, j] = dt * 0.0001  # scale factor


# Step 3: Create a dataframe with results (optional but nice)
dates = [f"{y}-{m:02d}" for (y, m) in map(extract_date, files_sorted)]
print(dates)
df = pd.DataFrame(sif_values, columns=['NE1', 'NE2', 'NE3', 'VAR'])
df['Date'] = pd.to_datetime(dates)
df = df.set_index('Date').sort_index()

# Show result
print(df.head())

#df.to_csv('GOSIF_timeseries.csv')
```
